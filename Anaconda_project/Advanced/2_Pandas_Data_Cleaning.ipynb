```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
meadian_salary_year = df['salary_year_avg'].median()
```

```python
meadian_salary_hour = df['salary_hour_avg'].median()
```

```python
#Create new data frame from the old one
df_filled = df

#Fill in the Nan values with the median. Set the filling to the column itself.
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(meadian_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(meadian_salary_hour)
```

```python
#Check the first rows
df_filled.loc[10:20, 'salary_year_avg':'salary_hour_avg']
```

```python
# Create new data frame to drop duplicates

df_unique = df_filled

df_unique = df_unique.drop_duplicates()

print('Length of original df:        ', len(df_filled))
print('Length pf drop duplicates df: ', len(df_unique))
print('Rows Dropped:                 ', len(df_filled) - len(df_unique))
```

```python
#Dropping Duplicates by using a subset
df_unique = df_unique.drop_duplicates(subset=['job_title','company_name'])

print('Length of original df:        ', len(df_filled))
print('Length pf drop duplicates df: ', len(df_unique))
print('Rows Dropped:                 ', len(df_filled) - len(df_unique))
```
