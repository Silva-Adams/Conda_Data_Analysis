```python
# Importing Libraries
from tqdm.notebook import trange, tqdm
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
#Reset_index()
#Filtering the df for USA jobs

df_usa = df[df['job_country']== 'South Africa']
```

```python
#Check the data frame. There are missing rows in the job index, since we filtered by country
df_usa.head()
```

# Correct the index step flow using reset_index method

```python

df_usa.reset_index(inplace=True)
df_usa
```

# Set Index Method

```python
df_usa.set_index('job_index', inplace=True)

df_usa
```

# Sort Index Method

```python
#Use the sort index in pivot tables or other functions
median_pivot = df_usa.pivot_table(values='salary_year_avg', index='job_title_short',aggfunc=['median','min','max'])

median_pivot
```

```python
median_pivot.sort_values(by=('median','salary_year_avg') ,ascending=False)
```

```python
median_pivot.sort_index(inplace=True)

median_pivot
```
